```python
#1. Librerías.
import pandas as pd
```

```python
#2. Constantes.
df_final_1_path = "./pruebas_batch/df_final_df_final_0_5000.csv"
df_final_2_path = "./pruebas_batch/df_final_df_final_5000_10000.csv"
df_final_3_path = "./pruebas_batch/df_final_df_final_10000_16000.csv"
df_final_4_path = "./pruebas_batch/df_final_df_final_reproceso_errores.csv"
path_df_final_exportacion = "./pruebas_batch/df_final_crudo.csv"
```

```python
#3. Lectura.
df_final_1 = pd.read_csv(df_final_1_path)
df_final_2 = pd.read_csv(df_final_2_path)
df_final_3 = pd.read_csv(df_final_3_path)
df_final_4 = pd.read_csv(df_final_4_path)
```

```python
#4. Concateno y reemplazo los batch que luego recuperé en el batch de errores.
#a. Unificamos los df principales en una lista.
dfs_originales = [df_final_1, df_final_2, df_final_3]

#b. Obtenemos los IDs reprocesados.
errores_ids = set(df_final_4["custom_id"].unique())

#c. Filtramos los originales, eliminando los que fueron reprocesados.
dfs_filtrados = []
for df in dfs_originales:
    df_filtrado = df[~df["custom_id"].isin(errores_ids)]
    dfs_filtrados.append(df_filtrado)

#d. Concatenamos todo: originales filtrados + reproceso.
df_total = pd.concat(dfs_filtrados + [df_final_4], ignore_index=True)

#e. Eliminamos la columna de custom_id.
del df_total["custom_id"]
#f. Borramos todas las noticias cuyos resultados están vacíos (9 casos), excluyendo las primeras 6 columnas.
df_total = df_total[~df_total.iloc[:, 6:].isna().all(axis=1)]
```

```python
#5. Verificamos duplicados.
duplicados = df_total.duplicated().sum()
print(f"🔎 Duplicados encontrados: {duplicados}.")
```

```python
#6. Exportamos.
df_total.to_csv(path_df_final_exportacion, index=False)
print(f"✅ df_final_crudo.csv creado con {len(df_total)} filas.")
```
